```python
from statsmodels.stats.proportion import proportions_ztest
```

```python
control_users = 4978
treatment_users = 5022

control_conversions = 2819
treatment_conversions = 2922
```

```python
successes = [control_conversions, treatment_conversions]
samples = [control_users, treatment_users]

print(successes)
print(samples)
```

```python
z_stat, p_value = proportions_ztest(successes, samples)

print(z_stat)
print(p_value)
```
